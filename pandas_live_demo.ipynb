# Pandas Live Demo: Data Engineering & Analytics


```python
import pandas as pd
import numpy as np
import seaborn as sns

pd.set_option("display.max_columns", None)

# Load open dataset
df = sns.load_dataset("titanic")
df.head()
```

## 1. Inspecting & Understanding the Data

```python
df.shape
```

```python
df.info()
```

```python
df.describe()
```

## 2. Missing Values

```python
df.isna().sum()
```

```python
df["age"].fillna(df["age"].median(), inplace=True)
df["age"].isna().sum()
```

## 3. Column Selection

```python
df["fare"].head()
```

```python
df[["sex", "age", "fare", "survived"]].head()
```

## 4. Creating a New DataFrame

```python
analytics_df = df[["sex", "class", "age", "fare", "survived"]]
analytics_df.head()
```

## 5. Filtering Rows

```python
df[df["fare"] > 100].head()
```

```python
df[(df["sex"] == "female") & (df["age"] < 30)].head()
```

## 6. Sorting

```python
df.sort_values("fare", ascending=False).head()
```

## 7. Grouping & Aggregation (OLAP-style)

```python
df.groupby("sex")["survived"].mean()
```

```python
df.groupby("class").agg(
    avg_fare=("fare", "mean"),
    median_age=("age", "median"),
    survival_rate=("survived", "mean")
)
```

## 8. Counting & Distribution

```python
df["class"].value_counts()
```

## 9. Renaming Columns

```python
df = df.rename(columns={"survived": "survival_flag"})
df.head()
```

## 10. Creating New Columns

```python
df["family_size"] = df["sibsp"] + df["parch"]
df[["sibsp", "parch", "family_size"]].head()
```

```python
df["fare_per_person"] = df["fare"] / (df["family_size"] + 1)
df[["fare", "family_size", "fare_per_person"]].head()
```

## 11. Applying Functions

```python
df["age_group"] = df["age"].apply(
    lambda x: "child" if x < 18 else "adult"
)
df[["age", "age_group"]].head()
```

# Pandas vs NumPy: Columnar vs Row-Based

This section demonstrates **why Pandas is column-oriented** and **NumPy is row-major**.


```python
# Convert to NumPy
arr = df[["age", "fare", "family_size"]].to_numpy()
arr[:5]
```

```python
# NumPy row access (fast)
arr[0, :]
```

```python
# NumPy column access (strided)
arr[:, 0][:5]
```

```python
# Pandas column-wise operation
df["fare"].mean()
```

```python
# NumPy equivalent
np.mean(arr[:, 1])
```

## Summary

**Pandas**
- Column-oriented
- Great for filtering, grouping, analytics
- OLAP-style workloads

**NumPy**
- Row-major memory layout
- Optimized for math and linear algebra
- Foundation for ML and scientific computing
