## Change Directories

In Collab, we need to use the % sign to physically change directories.

```python
%cd git_example_collab
```

## Create and Checkout a Branch

We title our branch feature/project-structure.

```python
!git checkout -b feature/project-structure
```

We can use !git branch in order to see which branch we are currently on, and how many there are.

```python
!git branch
```

## Make Directory Structure

```python
!mkdir data notebooks src tests docs
```

We 'touch' each directory and make placeholder files so that Git tracks the folders.

```python
!touch data/.gitkeep
!touch notebooks/example.ipynb
!touch src/__init__.py
!touch tests/__init__.py
!touch docs/README.md
```

Update the README.md file with a quick note.

```python
!echo "## Project Structure" >> README.md
```

List the directories to verify we created them.

```python
!ls
```

Verify there are changes being tracked using 'status'.

```python
!git status
```

## Stage Changes

We can stage changes using 'add'. By passing '.' we are saying add everything from where we are.

```python
!git add .
```

## Configure Username and Email

Git needs to know who you are when trying to make changes to a repo.

```python
  !git config --global user.email ""
  !git config --global user.name ""
```

## Commit the Changes

It is best practice to add a message that specifies what was changed.

```python
!git commit -m "Add standard project directory structure"
```

## Checkout Main

All of the work up to this point was completed on a branch. We now need to checkout main in order to merge our changes and updated main.

```python
!git checkout main
```

Again, verify which branch we are on.

```python
!git branch
```

## Merge to Main

```python
!git merge feature/project-structure
```

## Create Log File for Submission

'log' tracks changes that have been made. We can write the output to a text file.

```python
!git log --oneline --graph --decorate --all > git_activity.txt
```

Add a line to separate what we will add to the file.

```python
!echo -e "\n\n===== REPOSITORY STRUCTURE =====\n" >> git_activity.txt
```
