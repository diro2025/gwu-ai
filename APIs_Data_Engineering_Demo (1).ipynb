## 5. Simulated Pagination Loop
In real systems, you would loop over pages until no data remains.

```python
all_results = []
page = 1

while True:
    response = requests.get(
        "https://jsonplaceholder.typicode.com/posts",
        params={"_page": page, "_limit": 10}
    )

    page_data = response.json()

    if not page_data:
        break

    all_results.extend(page_data)
    page += 1

len(all_results)
```

## 7. Local Service API Concept

A local service API runs on your machine and is accessed via:

http://localhost:<port>

Example (if running a local server):

This is common in microservice architectures and local development environments.



### 7.1 Local Service API Demo (Flask)

We will:
1. Install Flask  
2. Create a small API  
3. Run it locally  
4. Call it using `requests`  

This simulates a microservice running on your machine.


```python

!pip install flask
```

```python
from flask import Flask, jsonify
import threading

app = Flask(__name__)

# Sample data
users = [
    {"id": 1, "name": "Alice"},
    {"id": 2, "name": "Bob"},
    {"id": 3, "name": "Charlie"}
]

@app.route("/users", methods=["GET"])
def get_users():
    return jsonify(users)

@app.route("/users/<int:user_id>", methods=["GET"])
def get_user(user_id):
    for user in users:
        if user["id"] == user_id:
            return jsonify(user)
    return jsonify({"error": "User not found"}), 404

def run_app():
    app.run(port=5000)

thread = threading.Thread(target=run_app)
thread.daemon = True
thread.start()
```

```python
# Now call our LOCAL API

import requests

response = requests.get("http://127.0.0.1:5000/users")
print("Status Code:", response.status_code)
print("Response JSON:", response.json())
```

```python
response = requests.get("http://127.0.0.1:5000/users/2")
print("Status Code:", response.status_code)
print("Response JSON:", response.json())
```

## 8. GraphQL Example

GraphQL uses a single endpoint and a query structure defined by the client.
We send a POST request containing the query as JSON.

Public GraphQL endpoint used:
https://countries.trevorblades.com/


```python
graphql_endpoint = "https://countries.trevorblades.com/"

continents_query = '''
{
  continents {
    code
    name
  }
}
'''

response = requests.post(graphql_endpoint, json={"query": continents_query})

continents = response.json()["data"]["continents"]

continents
```

```python
count_query = '''
{
  continents {
    name
    countries {
      code
    }
  }
}
'''

response = requests.post(graphql_endpoint, json={"query": count_query})

data = response.json()["data"]["continents"]

# Compute count per continent
continent_counts = [
    {"continent": c["name"], "country_count": len(c["countries"])}
    for c in data
]

continent_counts
```

```python
graphql_endpoint = "https://countries.trevorblades.com/"

query = '''
{
  countries(filter: { continent: { eq: "EU" } }) {
    code
    name
    capital
  }
}
'''

response = requests.post(graphql_endpoint, json={"query": query})

print("Status Code:", response.status_code)
```

```python
data = response.json()

# Inspect top-level keys
data.keys()
```

```python
countries = data["data"]["countries"]

len(countries), countries[:5]
```
